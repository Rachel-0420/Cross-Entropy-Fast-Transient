```python
import matplotlib.pyplot as plt
from matplotlib.patches import Rectangle, FancyArrowPatch

fig = plt.figure(figsize=(12,4))

# nightly triplet
ax1 = fig.add_axes([0.05, 0.18, 0.25, 0.68])
ax1.set_xlim(0, 10)
ax1.set_ylim(0, 4)
ax1.axis("off")

ax1.plot([1, 9], [2, 2], lw=1.5)
times = [2, 5, 8]
labels = [r"$m_{F_1}(t)$", r"$m_{F_2}(t+\Delta T_1)$", r"$m_{F_1}(t+\Delta T_2)$"]
for x, lab in zip(times, labels):
    ax1.plot(x, 2, 'o')
    ax1.text(x, 2.35, lab, ha='center', fontsize=10)

ax1.annotate(r"$\Delta T_1$", xy=(2,1.5), xytext=(5,1.5),
             arrowprops=dict(arrowstyle="<->"), ha='center')
ax1.annotate(r"$\Delta T_2$", xy=(2,0.9), xytext=(8,0.9),
             arrowprops=dict(arrowstyle="<->"), ha='center')

ax1.text(5, 3.3,
         r"$c = m_{F_1}(t)-m_{F_2}(t+\Delta T_1)$" "\n"
         r"$\Delta m = m_{F_1}(t)-m_{F_1}(t+\Delta T_2)$",
         ha='center', va='top', fontsize=10)
ax1.set_title("Nightly triplet")

# binning in (binMag, binColor) plane
ax2 = fig.add_axes([0.38, 0.18, 0.22, 0.68])
ax2.set_xlim(0, 6)
ax2.set_ylim(0, 6)
ax2.set_xlabel(r"$\Delta m$")
ax2.set_ylabel(r"$c$")

for x in range(6):
    ax2.axvline(x, lw=0.8, alpha=0.5)
for y in range(6):
    ax2.axhline(y, lw=0.8, alpha=0.5)

ax2.plot(3.4, 2.7, 'o')
ax2.add_patch(Rectangle((3,2), 1, 1, fill=False, lw=2))
ax2.text(3.5, 1.6, r"bin $(i,j)$", ha='center')
ax2.set_title(r"Binning in $(\Delta m, c)$")

# stack into hypercube
ax3 = fig.add_axes([0.69, 0.18, 0.25, 0.68])
ax3.set_xlim(0, 10)
ax3.set_ylim(0, 8)
ax3.axis("off")

# draw three stacked slices
for k, y in enumerate([1.2, 2.2, 3.2]):
    ax3.add_patch(Rectangle((2+y*0.2, y), 4, 2.2, fill=False, lw=1.2))
    ax3.text(7.2, y+1.1, rf"$(b,\Delta t)_{{{k+1}}}$", va='center')

ax3.text(4.5, 6.2, r"stack 2D histograms", ha='center')
ax3.text(4.5, 5.4, r"$n_{b,\Delta t,i,j}$", ha='center', fontsize=13)
ax3.add_patch(FancyArrowPatch((1,4), (2.1,4), arrowstyle='->', mutation_scale=12))
ax3.set_title("Presto-Color hypercube")

plt.show()
```

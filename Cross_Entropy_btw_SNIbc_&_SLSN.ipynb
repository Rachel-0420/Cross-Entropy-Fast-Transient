```python
with open(data1, 'rb') as f1:
    CoordinateDict1 = pickle.load(f1)
    Data1 = pickle.load(f1)

with open(data2, 'rb') as f2:
    CoordinateDict2 = pickle.load(f2)
    Data2 = pickle.load(f2)
```

```python
arr_2 = np.array(Data2, dtype=float)
shape = arr_2.shape
print(shape)
```

```python
# Change the data from int into float so that later we can change all the 0s to 1.e-16 to apply the cross entorpy equation to those data

Data1 = Data1.astype(float)
Data2 = Data2.astype(float)
```

```python
def normalize_for_4D_data(data, unit_1, unit_2, unit_3, unit_4):

    total = np.sum(data) * unit_1 * unit_2 * unit_3 * unit_4

    factor = 1 / total

    for i in range(shape[0]):
        for j in range(shape[1]):
            for k in range(shape[2]):
                for l in range(shape[3]):
                    data[i][j][k][l] *= factor

    return None
```

```python
unit_1 = 1
unit_2 = 20
unit_3 = 0.10
unit_4 = 0.50
```

```python
normalize_for_4D_data(Data1, unit_1, unit_2, unit_3, unit_4)
normalize_for_4D_data(Data2, unit_1, unit_2, unit_3, unit_4)
```

```python
# Check for normalization

print(math.isclose(np.sum(Data1), 1.))
print(math.isclose(np.sum(Data2), 1.))
```

```python
# Split to reduce use of RAM

part1 = Data2[:len(Data2)/2]
part2 = Data2[len(Data2)/2:]
```

```python
part1[np.where(part1 == 0.)] = 1.e-16
part2[np.where(part2 == 0.)] = 1.e-16
```

```python
for i in range(len(part1)):
  Data2[i] = part1[i]
  Data2[len(Data2)/2+i] = part2[i]
```

```python
# Caculation of cross entropy

H = - Data1 * np.log(Data2)
print(np.sum(H))
```
